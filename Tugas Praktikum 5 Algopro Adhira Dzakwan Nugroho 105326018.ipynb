```python
# Adhira Dzakwan Nugroho
# 105326018
# a. pengeluaran harian Nobita
hari = 1
saldo = 50000

while hari <= 30:
    spend = int(input(f"Hari ke-{hari}, masukkan pengeluaran: Rp "))
    saldo -= spend
    
    if saldo < 0:
        print("Saldo habis! Nobita tidak bisa lanjut jajan.")
        break
    
    print(f"Sisa saldo hari ke-{hari}: Rp {saldo}")
    hari += 1
else:
    print("Pencatatan selesai tanpa kehabisan saldo.")
```

```python
# b. menganalisis pengeluaran mingguan Nobita
weekly_spend = [20000, 0, 15000, 0]

for i in range(4):
    if weekly_spend[i] == 0:
        continue
    print(f"Pengeluaran minggu ke-{i+1}: Rp {weekly_spend[i]}")
else:
    print("✅ Analisis pengeluaran selesai.")
```

```python
# c. laporan bulanan Nobita
for bulan in range(1, 4):
    total_bulanan = 0
    print(f"\n📅 Bulan ke-{bulan}")
    
    for week in range(1, 5):
        spend = int(input(f"  Minggu ke-{week}, pengeluaran: Rp "))
        if spend > 20000:
            print("⚠️ Pengeluaran terlalu besar! Bulan dihentikan.")
            break
        total_bulanan += spend
    else:
        print(f"Total pengeluaran bulan ke-{bulan}: Rp {total_bulanan}")
    
print("\n🧾 Ringkasan pencatatan selesai.")
```

```python
# d. evaluasi akhir tahun
saldo_awal = 300000
saldo = saldo_awal

for bulan in range(1, 13):
    print(f"\n🗓️ Bulan ke-{bulan}")
    week = 1
    
    while week <= 4:
        libur = input(f"  Minggu ke-{week}, libur? (y/n): ").lower()
        if libur == 'y':
            week += 1
            continue
        spending = int(input("    Masukkan pengeluaran: Rp "))
        saldo -= spending
        if saldo < 0:
            print("💸 Saldo habis sebelum akhir tahun!")
            break
        week += 1
    else:
        print(f"Saldo akhir bulan ke-{bulan}: Rp {saldo}")
        continue
    break
else:
    print(f"\n🏆 Nobita berhasil menabung Rp {saldo} dari total Rp {saldo_awal}")
```
