```python
# Tugas Praktikum 6 Algopro Adhira Dzakwan Nugroho
# 105326018


skor_nobita = 0

def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai

def belajar(jam_belajar):
    if jam_belajar < 1:
        print("Nobita hanya membuka buku sebentar")
        return 50
    elif 1 <= jam_belajar < 3:
        print("Nobita belajar secukupnya")
        return 25
    else:
        print("Nobita belajar dengan rajin hari ini")
        return 10

def tidur(jam_tidur):
    if jam_tidur < 1:
        print("Nobita kurang tidur")
        return -10
    elif 1 <= jam_tidur < 2:
        print("Waktu tidur nobita seimbang")
        return 20
    else:
        print("Nobita tidur terlalu lama")
        return -20

def main_bersama_doraemon(jumlah_gadget, waktu_main):
    if waktu_main <= 1 and jumlah_gadget <= 2:
        print("Nobita bermain secukupnya")
        return 25
    elif waktu_main <= 2 and jumlah_gadget <= 3:
        print("Nobita masih bisa mengatur waktu dengan baik")
        return 5
    else:
        print("Nobita kebanyakan bermain")
        return -30

def menyontek():
    print("Nobita ketahuan menyontek")
    return -50

def jalankan_aktivitas(pilihan):
    if pilihan == 1:
        jam_belajar = float(input("Berapa jam Nobita belajar hari ini? "))
        skor = belajar(jam_belajar)
        update_skor(skor)
    elif pilihan == 2:
        jam_tidur = float(input("Berapa jam Nobita tidur siang hari ini? "))
        skor = tidur(jam_tidur)
        update_skor(skor)
    elif pilihan == 3:
        jumlah_gadget = int(input("Berapa gadget yang digunakan nobita? "))
        waktu_main = float(input("Berapa jam Nobita bermain gadget? "))
        skor = main_bersama_doraemon(jumlah_gadget, waktu_main)
        update_skor(skor)
    elif pilihan == 4:
        skor = menyontek()
        update_skor(skor)
    else:
        print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan")

def evaluasi_akhir():
    global skor_nobita
    if skor_nobita >= 80:
        return "Siswa Teladan"
    elif 50 <= skor_nobita < 80:
        return "Cukup Rajin"
    else:
        return "Pemalas"

def main():
    while True:
        print("\n=== PROGRAM PENINGKAT DISIPLIN NOBITA ===")
        print("1. Belajar")
        print("2. Tidur")
        print("3. Main bersama Doraemon")
        print("4. Menyontek saat ujian")
        print("5. Keluar & Lihat Evaluasi Akhir")
        print("=========================================")

        try:
            pilih = int(input("Pilih aktivitas (1-5): "))

            if pilih == 5:
                print("\n=========================================")
                print(f"Skor Akhir Disiplin Nobita: {skor_nobita}")
                print(f"Hasil Evaluasi Perilaku: {evaluasi_akhir()}")
                print("=========================================")
                print("Program Selesai. Terima Kasih")
                break

            jalankan_aktivitas(pilih)
            print(f"[Skor Saat Ini: {skor_nobita}]")

        except ValueError:
            print("Masukkan harus berupa angka!")

if __name__ == "__main__":
    main()



```
