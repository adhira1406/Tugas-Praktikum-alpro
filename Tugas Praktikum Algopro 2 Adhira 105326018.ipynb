```python
# Nama: Adhira Dzakwan NUgroho
# Tugas Praktikum Algopro 2

import math
# 1. Menyimpan data diri Giant ke dalam variabel
nama = "Giant"
usia = 20
berat_str = "97.5"
tinggi = 1.75
status_olahraga = False

# 2. Konversi berat badan menjadi float dan cetak kalimat pengantar
berat = float(berat_str)
print("Halo, saya {}. Usia saya {} tahun dan berat saya {} kg".format(nama, usia, berat))

# 3. Menghitung BMI (Cetak hasil asli dan bulatkan ke 2 desimal)
bmi_asli = berat/ (tinggi ** 2)
bmi_bulat = round(bmi_asli, 2)
print("BMI Asli: {}".format(bmi_asli))
print("BMI Dibulatkan (2 desimal): {}".format(bmi_bulat))

# 4. Perhitungan target minum air (Target: 2000ml, 1 botol = 350ml)
target_air = 2000
kapasitas_botol = 350
hasil_asli = target_air / kapasitas_botol # true division
botol_penuh = target_air // kapasitas_botol # floor division
sisa_air = target_air % kapasitas_botol # modulo
print("Hasil asli pembagian botol: {} botol".format(hasil_asli))
print("Jumlah botol penuh yang dibutuhkan: {} botol".format(botol_penuh))
print("Sisa air yang belum terwadahi botol penuh: {} ml".format(sisa_air))

# 5. Pembulatan kebutuhan kalori (2100.5 kalori)
kalori = 2100.5
kalori_bawah = math.floor(kalori)
kalori_atas = math.ceil(kalori)
print("Kalori dibulatkan ke bawah (floor): {}".format(kalori_bawah))
print("Kalori dibulatkan ke atas (ceil): {}".format(kalori_atas))



```
