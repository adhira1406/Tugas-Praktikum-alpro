```python
# Tugas Praktikum Algopro 3  
# Adhira Dzakwan Nugroho
# 105326018

# 1. Membuat Data Belanjaan
belanjaan = ["Apel", "Susu", "Roti"]
harga_jumlah = (5000, 3, 12000, 2, 8000, 1)
print("List belanjaan:", belanjaan)
print("Tuple harga jumlah:", harga_jumlah)
print("Karakter pertama:", belanjaan[0][0])
print("Karakter terakhir:", belanjaan[-1][-1])

# 2. Menghitung Total Harga Per Barang
total_apel = harga_jumlah[0] * harga_jumlah[1]
total_susu = harga_jumlah[2] * harga_jumlah[3]
total_roti = harga_jumlah[4] * harga_jumlah[5]
total_belanja = total_apel + total_susu + total_roti
print("\nTotal Apel:", total_apel)
print("Total Susu:", total_susu)
print("Total Roti:", total_roti)
print("Total Belanja:", total_belanja)

# 3. Mencetak Struk Belanja 
nama_pembeli = "Nobita Nobi"
nama_toko = "Toko Semnbako Doraemon"
print("\n" + "=" * 40)
print(f"{nama_toko:^40}")
print("=" * 40)
print(f"pembeli: {nama_pembeli}")
print(f"{'Barang':<12}{'Harga':>8}{'Jumlah':>8}{'Total':>10}")
print("-" * 40)
for i in range(3):
    harga = harga_jumlah[i * 2]
    jumlah = harga_jumlah[i * 2 * 1]
    total = harga * jumlah
    print(f"{belanjaan[i]:<12}{harga:>8}{jumlah:>8}{total:>10}")
print("-" * 40)
print(f"{'Total Belanja':<28}{total_belanja:>12}")
print(f"{'Terima kasih sudah belanja!':^40}")    

# 4. Menambahkan Barang Baru 
belanjaan.append("Pisang")
print("\nList belanjaan terbaru:", belanjaan)
```
