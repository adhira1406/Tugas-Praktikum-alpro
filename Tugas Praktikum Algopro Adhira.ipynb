```python
# Tugas Praktikum 4 Algopro 
# Adhira Dzakwan Nugroho (105326018)

# ====================================
# Vehicle Insurance Analyzer
# ====================================

# 1. Menerima Input dari Pengguna  
usia_pengemudi = int(input("1. Masukkan usia pengemudi (tahun): "))
jenis_kendaraan = input("2. Masukkan jenis kendaraan (mobil/motor): ").lower()
nilai_kendaraan = float(input("3. Masukkan nilai kendaraan (rupiah): "))
area_penggunaan = input("4. Masukkan area penggunaan (perkotaan/pedesaan): ").lower()
sistem_pengaman = input("5. Apakah memiliki sistem pengaman (ya/tidak): ").lower()

# 2. Menentukan Risiko Awal berdasarkan Jenis Kendaraan
if jenis_kendaraan == "motor":
    risiko = "sedang"
else:
    risiko = "rendah"

# 3. Validasi Risiko berdasarkan Usia Pengemudi 
if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko : "tinggi"

# 4. Validasi Risiko berdasarkan Area Penggunaan 
if area_penggunaan == "perkotaan":
    if risiko == "rendah":
        risiko = "sedang"
    elif risiko == "sedang":
        risiko = "tinggi"

# 5. Validasi Risiko berdasarkan Sistem Pengaman 
if sistem_pengaman == "tidak":
    if risiko == "rendah":
        risiko = "sedang"
    elif risiko == "sedang":
        risiko = "tinggi"
    else:
        risiko = "tinggi"

# 6. Menghitung Premiu Kendaraan berdasarkan Risiko Akhir  
if risiko == "tinggi":
    persentase_premi = 0.1
elif risiko == "sedang":
    persentase_premi = 0.07
else:
    persentase_premi = 0.05
premi_dibayar = persentase_premi * nilai_kendaraan

# 7. Menentukan Pesan Doraemon 
pesan_doraemon = (
    "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!"
    if risiko == "tinggi"
    else "Risiko kendaraan masih aman menurut Doraemon"
)

# 8. Menampilkan Output
print("\n===========================================")
print("Ekspetasi Output:")
print("===========================================")
print(f"1. Kategori risiko: {risiko.upper()}")
print(f"2. Premi asuransi yang perlu dibayarkan: Rp {premi_dibayar:,.2f}") 
print(f"3. Pesan Doraemon: {pesan_doraemon}")
print("===========================================")
```
